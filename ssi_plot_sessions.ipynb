```python
from SessionData import SessionData
from behavior_utils import *
from ssi_plot import *
```

```python
# Use event.csv to find valid sessions
base_path    = "d:/data/"
experiment   = "clickbait-motivate"
event_paths = get_file_paths(f"{base_path}/{experiment}/events/", 'csv', '', session_type='', print_paths=True, print_n=3)
```

```python
# Gather mouse and session IDs from event paths
mouse_ids = []
session_ids = []

for event_path in event_paths:
    mouse_id = event_path.parts[-3]     
    session_id = event_path.parts[-2]
    mouse_ids.append(mouse_id)
    session_ids.append(session_id)
```

```python
for ii, event_path in enumerate(event_paths):
    print(f"{ii}: {event_path}")
```

```python
# Permutation count for spatial significance test
n_perms = 1000

# Iterate over sessions and run SSI analysis
for ii in range(len(mouse_ids)):
    mouse_id = mouse_ids[ii]
    session_id = session_ids[ii]
    print(f"Loading data: mouse {mouse_id} / session {session_id}")
    try:
        data = SessionData(
            base_path=base_path,
            mouse_id=mouse_id,
            session_id=session_id,
            experiment=experiment,
            min_spikes=50,
            verbose=True,
        )
        print(f"{data.n_clusters} clusters loaded.\n")

        records = run_session_ssi(
            data, mouse_id, session_id,
            grid_shape=(5, 11),
            min_occupancy_ms=2000,
            n_permutations=n_perms,
        )

        print(f"\nDone. {len(records)} clusters processed.")
    except Exception as e:
        print(f"Error processing mouse {mouse_id} / session {session_id}: {e}")
```
